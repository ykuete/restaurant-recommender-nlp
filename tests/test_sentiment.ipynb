```python
import joblib
```

```python
model = joblib.load("../models/sentiment_logistic_model.joblib")
tfidf = joblib.load("../models/tfidf_vectorizer.joblib")

example = ["Yay!! Taste So disgustingly awful"]

example_vec = tfidf.transform(example)

pred = model.predict(example_vec)

print(pred)
```
