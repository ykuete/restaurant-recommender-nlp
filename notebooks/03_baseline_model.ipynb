## **Baseline Model: Popularity Recommendation Engine**

Display top 10 highest rated restaurant in Philadelphia

```python
def compute_wr(x, m, C):            # Function to compute the weighted rating for restaurant based on existing dataset
    v = x['review_count']
    r = x['stars']
    score = (v/(v+m))*r + (m/(v+m))*C
    return score
```

```python
C = phila_res_df['stars'].mean()
m = phila_res_df['review_count'].quantile(0.90) 
new_df = phila_res_df.copy().loc[phila_res_df['review_count'] >= m]
new_df['weight_score'] = compute_wr(new_df, m, C)
new_df
```

**I picked out the 90% upper quantile of the data to filter restaurants that have too few reviews to be useful**

- Since the dataset contains restaurant with a 5 stars reviews, it is realistically unlikely. But there are several reasons as to why this happens:
    1. The restaurant is too new or has too few reviews to best represent the establishment
    2. The restaurents' reviews are likely fabricated
- Either way, restaurants with too few reviews is not likely to yield useful insights
- From a specific set of restaurants, I computed the weight rating of the entire dataset to return a the best restaurants in the data.

```python
baseline_recommendation = new_df.sort_values("weight_score", ascending= False)
baseline_recommendation.head(10)
```

The output above returns top 10 best restaurant in Philadelphia, with their respective weight score on the far right. As we can see, these top restaurants are not tailored or personalized to best represent a user's interest, however this recommendation would be a greate baseline to measure other semantic and sentimental model againts.
